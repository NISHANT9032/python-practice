```python
import requests
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client()

url = "https://api.frankfurter.app/latest?from=USD&to=INR"

response = requests.get(url)

if response.status_code == 200:
    rate = response.json()["rates"]["INR"]

    print(f"Current USD to INR Rate: {rate}")

    prompt = f"""
    I have 200 NVIDIA stocks.
    Today's price per stock is $215.

    Calculate the total value in USD and convert it to Indian rupees.

    Today's USD to INR conversion rate is {rate}.

    Provide a clear calculation.
    """

    try:
        answer = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=prompt
        ).text

        print("\nGemini Response:")
        print(answer)

    except Exception as e:
        print(f"Something went wrong with Gemini: {e}")

else:
    print("Failed to fetch exchange rate.")
```
