```python
def generate_recipe(ingredients, cuisine, diet):
    """
    Generate a recipe using the ingredients provided.

    Cuisine can be "Italian", "Mexican", "Indian", "Any", etc.
    Diet can be "Vegan", "Vegetarian", "Jain", "Any", etc.
    """

    prompt = f"""
    Generate one food recipe using these ingredients:
    {", ".join(ingredients)}

    Recipe should not be more than 100 words.

    Cuisine: {cuisine}
    Diet: {diet}
    """

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    return response.text
```

```python
recipe = generate_recipe(
    ["Tomatoes", "lentils", "onion", "jeera", "salt"],
    "Indian",
    "Vegan"
)

recipe
```

```python
from google import genai
from google.genai import types

# Create the Gemini client
client = genai.Client()

# Conversation memory
conversation = []

# Bot personality
config = types.GenerateContentConfig(
    system_instruction=(
        "You are a witty movie buff who loves making puns. "
        "Remember the user's movie preferences from the conversation "
        "and use them when recommending movies."
    )
)


def chat(user_message):
    # Add user's message to memory
    conversation.append(
        types.Content(
            role="user",
            parts=[types.Part(text=user_message)]
        )
    )

    # Send the entire conversation so far
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=conversation,
        config=config
    )

    # Get model response
    assistant_message = response.text

    # Add model response to memory
    conversation.append(
        types.Content(
            role="model",
            parts=[types.Part(text=assistant_message)]
        )
    )

    return assistant_message


def reset():
    """Start a new conversation by clearing memory."""
    conversation.clear()
    print("🎬 New chat started! I've forgotten our previous movie preferences.")
```
